En esta parte se resuelven los puntos del PDF del parcial. Todos los cálculos se hacen con `groupby`: se agrupa a las personas (por departamento, sexo, edad u oficio) y se suma o se promedia dentro de cada grupo.

Una aclaración que sirve para todo el parcial: la GEIH es una encuesta a una muestra de hogares, y cada persona encuestada representa a varias personas del país. Esa cantidad es el factor de expansión `FEX_C18`. Por eso, para **contar personas se suma `FEX_C18`** y no se cuentan filas.

> Para el total nacional calcule: población en edad de trabajar, población que no está en edad de trabajar, población económicamente activa, población económicamente inactiva, tasa de ocupación y tasa de desempleo.

```python
PT     = cgse['FEX_C18'].sum()                            # población total
PET    = ft.groupby('PET')['FEX_C18'].sum().loc[1]        # población en edad de trabajar (15 años o más)
NO_PET = PT - PET                                         # población que no está en edad de trabajar
PEA    = ft.groupby('FT')['FEX_C18'].sum().loc[1]         # población económicamente activa
PEI    = PET - PEA                                        # población económicamente inactiva
D      = desempleados.groupby('DSI')['FEX_C18'].sum().loc[1]   # desempleados
O      = ocupados['FEX_C18'].sum()                        # ocupados

TO = O / PET * 100     # tasa de ocupación = ocupados / población en edad de trabajar
TD = D / PEA * 100     # tasa de desempleo = desempleados / población económicamente activa

print(f'Población en edad de trabajar (PET):        {PET:,.0f}')
print(f'Población que NO está en edad de trabajar:  {NO_PET:,.0f}')
print(f'Población económicamente activa (PEA):      {PEA:,.0f}')
print(f'Población económicamente inactiva (PEI):    {PEI:,.0f}')
print(f'Ocupados: {O:,.0f}   |   Desempleados: {D:,.0f}')
print(f'Tasa de ocupación:  {TO:.2f} %')
print(f'Tasa de desempleo:  {TD:.2f} %')
```

La tasa de ocupación es de **59,36 %**: de cada 100 personas en edad de trabajar, unas 59 tienen empleo. La tasa de desempleo es de **8,03 %**: de cada 100 personas que quieren trabajar, unas 8 no consiguen empleo. Este dato coincide con la cifra oficial del DANE (8,0 %), lo que indica que el cálculo está bien.

## Pregunta 2 — Tasa de desempleo de hombres y mujeres por departamento
> Calcule las tasas de desempleo para hombres y mujeres en todos los departamentos de Colombia a Junio de 2026, ¿cuáles son los departamentos con mayores y menores tasas de desempleo?

Se agrupa por departamento y sexo: se suman los desempleados y la PEA de cada grupo y se divide (tasa = desempleados / PEA × 100).

```python
des_ds = des.groupby(['DPTO', 'P3271'])['FEX_C18'].sum()     # desempleados por departamento y sexo
pea_ds = pea.groupby(['DPTO', 'P3271'])['FEX_C18'].sum()     # PEA por departamento y sexo

td = des_ds.reindex(pea_ds.index, fill_value=0) / pea_ds * 100    # tasa de desempleo (si no hay desempleados, 0)
td = td.unstack('P3271')
td.columns = ['Hombres', 'Mujeres']
td['Total'] = des.groupby('DPTO')['FEX_C18'].sum().reindex(td.index, fill_value=0) / pea.groupby('DPTO')['FEX_C18'].sum() * 100
td.index = td.index.map(dpto_nombre)
td = td.sort_values('Total', ascending=False)

print('Tasa de desempleo por departamento (ordenados de mayor a menor desempleo total):\n')
for dpto, f in td.iterrows():
    print(f"{dpto:<20} hombres {f['Hombres']:5.1f} %    mujeres {f['Mujeres']:5.1f} %    total {f['Total']:5.1f} %")

print()
for col in ['Hombres', 'Mujeres', 'Total']:
    s = td[col].sort_values(ascending=False)
    print(f'{col} - mayores tasas: ' + ', '.join(f'{d} ({v:.1f} %)' for d, v in s.head(3).items()))
    print(f'{col} - menores tasas: ' + ', '.join(f'{d} ({v:.1f} %)' for d, v in s.tail(3).sort_values().items()))
print(f"\nDepartamentos donde las mujeres tienen más desempleo que los hombres: {(td['Mujeres'] > td['Hombres']).sum()} de {len(td)}")
```

**Respuesta.** El departamento con mayor desempleo es **Putumayo** (22,2 %; 19,7 % en hombres y 25,0 % en mujeres), seguido de San Andrés (19,7 %) y Vichada (15,6 %). Por sexo, los más altos en hombres son San Andrés (21,0 %), Putumayo (19,7 %) y Arauca (14,5 %), y en mujeres Putumayo (25,0 %), Sucre (20,5 %) y Vichada (18,4 %).

Los más bajos son **Vaupés** (2,5 %), Huila y Cauca (5,6 %). En hombres: Vaupés y Cauca (3,0 %) y Risaralda (4,0 %); en mujeres: Vaupés (1,8 %), Amazonas (2,6 %) y Huila (6,6 %). Hay que tener cuidado con Vaupés y Amazonas, porque tienen muy pocas encuestas y sus tasas son poco confiables.

Además, en **28 de los 33 departamentos las mujeres tienen más desempleo que los hombres**. La mayor diferencia está en Sucre (6,3 % hombres frente a 20,5 % mujeres). En el Quindío el desempleo es de 6,0 % en hombres y 7,3 % en mujeres, por debajo del promedio nacional.

## Pregunta 3 — Grupo de edad con mayor y menor desempleo en cada departamento
> Tomando en consideración únicamente los grupos de edad 18 a 24, 24 a 30, 30 a 40, 40 a 50 y 60 en adelante: para cada uno de los departamentos, ¿cuál es el grupo de edad con mayor y menor tasa de desempleo?

Se agrupa por departamento y grupo de edad, se calcula la tasa de cada grupo y se busca en cada departamento el grupo con la tasa más alta y el de la más baja.

```python
pea_e = pea.assign(Grupo=pea['P6040'].map(grupo_edad)).dropna(subset=['Grupo'])
des_e = des.assign(Grupo=des['P6040'].map(grupo_edad)).dropna(subset=['Grupo'])

pea_dg = pea_e.groupby(['DPTO', 'Grupo'])['FEX_C18'].sum()
des_dg = des_e.groupby(['DPTO', 'Grupo'])['FEX_C18'].sum()
td_edad = (des_dg.reindex(pea_dg.index, fill_value=0) / pea_dg * 100).unstack('Grupo')[ORDEN_GRUPOS]
td_edad.index = td_edad.index.map(dpto_nombre)

def extremo(fila, modo):
    fila = fila.dropna()
    valor = fila.max() if modo == 'max' else fila.min()
    return ' / '.join(fila[fila == valor].index), valor      # si hay empate, se muestran los dos grupos

mayores, menores = [], []
for dpto, fila in td_edad.iterrows():
    g_max, v_max = extremo(fila, 'max')
    g_min, v_min = extremo(fila, 'min')
    mayores.append(g_max); menores.append(g_min)
    print(f'{dpto:<20} mayor desempleo: {g_max} ({v_max:.1f} %)    menor desempleo: {g_min} ({v_min:.1f} %)')

print()
print('Grupo con MAYOR desempleo (número de departamentos): ' + ', '.join(f'{g}: {n}' for g, n in pd.Series(mayores).value_counts().items()))
print('Grupo con MENOR desempleo (número de departamentos): ' + ', '.join(f'{g}: {n}' for g, n in pd.Series(menores).value_counts().items()))
```

**Respuesta.** En casi todos los departamentos el desempleo es mayor entre los **jóvenes de 18 a 24 años** (25 de los 33 departamentos). En 7 departamentos es mayor en el grupo de 24 a 30 (Boyacá, Chocó, Nariño, Arauca, Putumayo, Amazonas y Vaupés) y en La Guajira es mayor en el de 40 a 50. El grupo con **menor desempleo es el de 60 años en adelante** en 24 departamentos (y empata como el menor en 2 más). Las excepciones son Boyacá, Nariño y San Andrés (40 a 50), Córdoba y Valle del Cauca (30 a 40), Guaviare (24 a 30) y Putumayo (18 a 24). El detalle de cada departamento está en el listado de arriba; por ejemplo, en el Quindío el mayor desempleo es de 18 a 24 años (15,6 %) y el menor de 60 en adelante (3,1 %).

Esto tiene sentido económicamente: los jóvenes buscan su primer empleo sin experiencia, mientras que quienes siguen activos después de los 60 casi siempre ya tienen trabajo o negocio propio, y quienes no, se pensionan y salen del mercado laboral. En departamentos pequeños (Vaupés, Guainía, Amazonas) hay muy pocas encuestas por grupo, por eso aparecen tasas de 0 %.

Se cuenta también cuántas mujeres encuestadas hay en cada departamento, porque con muy pocas la tasa no es confiable. (En la guía de arriba este cálculo no filtraba por sexo; aquí sí.)

```python
def mujeres_18_25(df):
    return df[(df['P3271'] == 2) & (df['P6040'] >= 18) & (df['P6040'] < 25)]

pea_m = mujeres_18_25(pea)
des_m = mujeres_18_25(des)

pea_d = pea_m.groupby('DPTO')['FEX_C18'].sum()                       # PEA de mujeres jóvenes por departamento
des_d = des_m.groupby('DPTO')['FEX_C18'].sum().reindex(pea_d.index, fill_value=0)
muestra = pea_m.groupby('DPTO').size()                               # mujeres encuestadas por departamento

td4 = (des_d / pea_d * 100).sort_values()
td4.index = td4.index.map(dpto_nombre); muestra.index = muestra.index.map(dpto_nombre)

print('Los 3 departamentos con menor desempleo de mujeres de 18 a 25 años:')
for i, (dpto, v) in enumerate(td4.head(3).items(), 1):
    print(f'  {i}. {dpto}: {v:.1f} %  ({muestra[dpto]} mujeres encuestadas)')

print('\nSi solo se cuentan departamentos con al menos 30 mujeres encuestadas:')
td4_ok = td4[muestra.reindex(td4.index) >= 30]
for i, (dpto, v) in enumerate(td4_ok.head(3).items(), 1):
    print(f'  {i}. {dpto}: {v:.1f} %  ({muestra[dpto]} mujeres encuestadas)')
```

Solo se incluyen los ocupados que reportan un ingreso mayor que cero (algunos no tienen dato o tienen 0, por ejemplo quienes trabajan en un negocio familiar sin sueldo). Se presentan dos promedios: el **simple** (promedio directo de lo que respondieron las personas encuestadas) y el **ponderado** (que tiene en cuenta a cuántas personas representa cada encuestado, y por eso es el más representativo del departamento).

```python
print('Ocupados sin dato de ingreso:', ocupados['INGLABO'].isna().sum(), '| con ingreso igual a 0:', (ocupados['INGLABO'] == 0).sum())

oc = ocupados[ocupados['INGLABO'] > 0].copy()           # ocupados con ingreso reportado
oc['INGRESO_x_FEX'] = oc['INGLABO'] * oc['FEX_C18']

def promedio_pond(df, por):
    """Promedio ponderado por grupo = suma(ingreso * factor de expansión) / suma(factor de expansión)."""
    g = df.groupby(por)
    return g['INGRESO_x_FEX'].sum() / g['FEX_C18'].sum()

oc_5 = oc[oc['DPTO'].isin(cinco)]
pond5  = promedio_pond(oc_5, 'DPTO')
simple5 = oc_5.groupby('DPTO')['INGLABO'].mean()
n5 = oc_5.groupby('DPTO').size()

print('\nIngreso laboral mensual promedio (de mayor a menor):')
for dpto in pond5.sort_values(ascending=False).index:
    print(f'{dpto_nombre[dpto]:<17} ponderado ${pond5[dpto]:>12,.0f}   simple ${simple5[dpto]:>12,.0f}   ({n5[dpto]:,} personas encuestadas)')
print(f"Los 5 juntos      ponderado ${oc_5['INGRESO_x_FEX'].sum() / oc_5['FEX_C18'].sum():>12,.0f}   simple ${oc_5['INGLABO'].mean():>12,.0f}")
```

**Respuesta.** El ingreso laboral promedio más alto es el de **Antioquia** (cerca de $2,49 millones al mes, promedio ponderado) y el más bajo el de **Valle del Cauca** (cerca de $2,15 millones). En el medio quedan Risaralda ($2,23 millones), Caldas ($2,23 millones) y Quindío ($2,17 millones). El promedio de los cinco departamentos juntos es de unos $2,33 millones.

En el Quindío una persona ocupada gana en promedio unos $315.000 menos que en Antioquia. Con el promedio simple el orden es parecido (Antioquia primero y Valle último), aunque ahí Caldas queda de segundo. Los promedios son altos en parte porque unas pocas personas ganan sumas muy grandes (hay ingresos de hasta 100 millones), lo que sube el promedio por encima de lo que gana una persona típica.
